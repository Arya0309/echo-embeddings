```python
import stanza
```

```python
nlp = stanza.Pipeline("en", processors="tokenize,pos,depparse")
doc = nlp(
    "She loves to travel in summer, especially to cold destinations, avoiding hot and crowded places."
)  # doc is class Document
dicts = doc.to_dict()
```

```python
from stanza.utils.conll import CoNLL
import pandas as pd

conll = CoNLL.convert_dict(dicts)
df = pd.DataFrame(conll[0])
df
```
